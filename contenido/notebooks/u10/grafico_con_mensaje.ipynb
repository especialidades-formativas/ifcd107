## 10.4 · Storytelling con datos

`a1_historia.py`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from banco import X_e, X_p, y_e, y_p
from evaluar import potenciacion

p = potenciacion().fit(X_e, y_e).predict_proba(X_p)[:, 1]
dec = pd.qcut(-p, 10, labels=range(1, 11))
tasa = 100 * y_p.groupby(dec.to_numpy()).mean()
color = ["C0" if d <= 2 else "lightgray" for d in tasa.index]
fig, ej = plt.subplots(figsize=(4.4, 2.2), layout="constrained")
ej.bar(tasa.index, tasa.values, color=color)
ej.axhline(100 * y_p.mean(), color="gray", ls=":", lw=0.8)
ej.text(10.4, 100 * y_p.mean() + 1, "media", ha="right",
        fontsize=6, color="gray")
ej.set_title("Llamar a 2 de cada 10 clientes consigue"
             f" el {100 * y_p[dec <= 2].sum() / y_p.sum():.0f} %"
             " de los depósitos", fontsize=7, loc="left")
ej.set_xlabel("decil según el modelo (1 = más prometedores)")
ej.set_ylabel("% que suscribe"); ej.set_xticks(range(1, 11))
for s in ["top", "right"]:
    ej.spines[s].set_visible(False)
plt.show()
print("tasas por decil:", tasa.round(1).tolist())
```
